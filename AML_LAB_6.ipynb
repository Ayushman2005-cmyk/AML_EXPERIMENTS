```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import chi2_contingency
```

```python
data = [[10, 5],
        [5, 10]]
df = pd.DataFrame(
    data,
    index=['Like Sports: Yes', 'Like Sports: No'],
    columns=['Buys: Yes', 'Buys: No']
)
print("Contingency Table:")
display(df)
chi2, p, dof, expected = chi2_contingency(
    data,
    correction=False
)
alpha = 0.05
tabulated_chi2 = 3.841
print(f"\nCalculated Chi-square statistic: {chi2:.4f}")
print(f"Tabulated Chi-square (alpha=0.05, df=1): {tabulated_chi2}")
print(f"P-value: {p:.4f}")
print(f"Degrees of freedom: {dof}")
expected_df = pd.DataFrame(
    expected,
    index=df.index,
    columns=df.columns
)
print("\nExpected Frequencies:")
display(expected_df)
if chi2 > tabulated_chi2:
    print("\nResult: Reject the null hypothesis")
    print("Significant relationship exists between Sports and buying protein powder.")
else:
    print("\nResult: Fail to reject the null hypothesis")
    print("No significant relationship exists between Sports and buying protein powder.")
```

```python
labels = ['Calculated Chi-square', 'Tabulated Chi-square (Critical Value)']
values = [chi2, tabulated_chi2]
colors = ['blue', 'green']
plt.figure(figsize=(8, 6))
plt.bar(labels, values, color=colors)
plt.ylabel('Chi-square Value')
plt.title('Comparison of Calculated vs Tabulated Chi-square')
plt.axhline(y=tabulated_chi2, color='r', linestyle='--', label='Significance Threshold')
for i, v in enumerate(values):
    plt.text(i, v + 0.05, f"{v:.3f}", ha='center', fontweight='bold')
plt.legend()
plt.grid()
plt.show()
```

```python
import math
from scipy import stats
```

```python
mu_0 = 100
sigma = 12
n = 36
x_bar = 104
alpha = 0.05
```

```python
std_error = sigma/math.sqrt(n)
```

```python
z_score = (x_bar - mu_0) / std_error
```

```python
p_value = 2 * (1 - stats.norm.cdf(abs(z_score)))
```

```python
z_critical = stats.norm.ppf(1 - alpha / 2)
```

```python
print(f"Standard Error: {std_error}")
print(f"Calculated Z-statistic: {z_score:.4f}")
print(f"Critical Z-value (two-tailed): +/-{z_critical:.4f}")
print(f"P-value: {p_value:.4f}")
```

```python
if abs(z_score) > z_critical:
    print("\nDecision: Reject the Null Hypothesis (H0).")
    print("Conclusion: There is sufficient evidence at the 5% significance level to suggest that the true mean processing time is different from 100ms.")
else:
    print("\nDecision: Fail to reject the Null Hypothesis (H0).")
    print("Conclusion: There is not enough evidence at the 5% significance level to suggest that the true mean processing time is different from 100ms.")
```

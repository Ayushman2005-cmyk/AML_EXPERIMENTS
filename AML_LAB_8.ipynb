```python
import pandas as pd
data = {
    'Obs': [1, 2, 3, 4, 5, 6],
    'Rainfall': ['H', 'H', 'H', 'L', 'L', 'L'],
    'Soiltype': ['Loamy', 'Sandy', 'Clay', 'Loamy', 'Sandy', 'Clay'],
    'Yield': ['Good', 'Good', 'Good', 'Poor', 'Poor', 'Good']
}
df = pd.DataFrame(data)
display(df)
```

```python
def evaluate_one_r(df, feature, target):
    print(f"=== Evaluating Feature: {feature} ===")
    freq_table = pd.crosstab(df[feature], df[target])
    print("Frequency Table:")
    display(freq_table)
    rules = {}
    total_correct = 0
    total_records = len(df)
    for value in freq_table.index:
        best_class = freq_table.loc[value].idxmax()
        correct_count = freq_table.loc[value, best_class]
        rules[value] = best_class
        total_correct += correct_count
        print(f"  Rule: If {feature} == {value} -> Predict {best_class} (Correct: {correct_count}/{freq_table.loc[value].sum()})")
    accuracy = total_correct / total_records
    error_rate = 1.0 - accuracy
    print(f"Total Accuracy: {accuracy:.2%} | Total Error Rate: {error_rate:.2%}\n")
    return rules, accuracy
rainfall_rules, rainfall_acc = evaluate_one_r(df, 'Rainfall', 'Yield')
soiltype_rules, soil_acc = evaluate_one_r(df, 'Soiltype', 'Yield')
if rainfall_acc > soil_acc:
    best_feature = 'Rainfall'
    best_acc = rainfall_acc
elif soil_acc > rainfall_acc:
    best_feature = 'Soiltype'
    best_acc = soil_acc
else:
    best_feature = 'Both features (Tie)'
    best_acc = rainfall_acc
print(f"Conclusion: The best single classifier is '{best_feature}' with an accuracy of {best_acc:.2%}.")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
sns.countplot(ax=axes[0], data=df, x='Rainfall', hue='Yield', palette='viridis')
axes[0].set_title('Yield Distribution by Rainfall')
axes[0].set_xlabel('Rainfall')
axes[0].set_ylabel('Count')
sns.countplot(ax=axes[1], data=df, x='Soiltype', hue='Yield', palette='viridis')
axes[1].set_title('Yield Distribution by Soiltype')
axes[1].set_xlabel('Soil Type')
axes[1].set_ylabel('Count')
accuracies = {'Rainfall': rainfall_acc, 'Soiltype': soil_acc}
sns.barplot(ax=axes[2], x=list(accuracies.keys()), y=list(accuracies.values()), palette='mako')
axes[2].set_title('One-R Classifier Accuracy')
axes[2].set_xlabel('Feature')
axes[2].set_ylabel('Accuracy')
axes[2].set_ylim(0, 1.1)
for index, value in enumerate(accuracies.values()):
    axes[2].text(index, value + 0.02, f"{value:.1%}", ha='center', fontweight='bold')
plt.tight_layout()
plt.show()
```

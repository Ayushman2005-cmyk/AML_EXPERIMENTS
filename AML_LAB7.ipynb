```python
import numpy as np
import pandas as pd
from scipy import stats

# Data initialization
univariate = np.array([82, 84, 81, 83, 85])
chi_square = np.array([85, 87, 86, 84, 88])
mutual_info = np.array([88, 89, 91, 90, 92])

data = [univariate, chi_square, mutual_info]
labels = ['Univariate', 'Chi-square', 'Mutual Information']

# 2. Calculate mean accuracy for each method
means = [np.mean(group) for group in data]
for label, mean in zip(labels, means):
    print(f"Mean accuracy for {label}: {mean:.2f}%")

# 3. Calculate the Grand Mean
all_data = np.concatenate(data)
grand_mean = np.mean(all_data)
print(f"Grand Mean: {grand_mean:.2f}%")

# 4. Calculate Sum of Squares
n_groups = len(data)
n_per_group = len(univariate)

# SSB (Between-Group)
ssb = sum(n_per_group * (m - grand_mean)**2 for m in means)

# SSW (Within-Group)
ssw = sum(sum((x - m)**2 for x in group) for group, m in zip(data, means))

print(f"SSB (Between-Group): {ssb:.2f}")
print(f"SSW (Within-Group): {ssw:.2f}")

# 5. Calculate Mean Squares
df_between = n_groups - 1
df_within = len(all_data) - n_groups

msb = ssb / df_between
msw = ssw / df_within

print(f"MSB: {msb:.2f}")
print(f"MSW: {msw:.2f}")

# 6. Calculate F-Statistic
f_stat = msb / msw
print(f"Calculated F-Statistic: {f_stat:.4f}")
# 7. Determine F critical value at alpha=0.05
alpha = 0.05
f_critical = stats.f.ppf(1 - alpha, df_between, df_within)
print(f"F critical value (alpha={alpha}, dfn={df_between}, dfd={df_within}): {f_critical:.4f}")

# 8. Decision
if f_stat > f_critical:
    decision = "Reject the Null Hypothesis (H0)"
else:
    decision = "Fail to Reject the Null Hypothesis (H0)"

print(f"Decision: {decision}")
```

```python
import numpy as np
import pandas as pd
from scipy.stats import chi2_contingency, chi2

# Observed frequencies
# Rows: High Accuracy, Low Accuracy
# Columns: Chi-Square, ANOVA, Mutual Information
observed = np.array([
    [18, 14, 20],  # High Accuracy
    [7, 11, 5]     # Low Accuracy
])

# 2. Construct the expected frequency table
chi2_stat, p_val, dof, expected = chi2_contingency(observed)

expected_df = pd.DataFrame(expected,
                           index=['High Accuracy', 'Low Accuracy'],
                           columns=['Chi-Square', 'ANOVA', 'Mutual Information'])

print("2. Expected Frequency Table:")
display(expected_df)

# 3. Calculate the chi-square statistic
print(f"\n3. Chi-square Statistic: {chi2_stat:.4f}")

# 4. Determine degrees of freedom
print(f"4. Degrees of Freedom: {dof}")

# 5. Find the critical chi-square value at alpha=0.05
alpha = 0.05
critical_value = chi2.ppf(1 - alpha, dof)
print(f"5. Critical Chi-square value at alpha={alpha}: {critical_value:.4f}")

# 6. Compare the calculated and critical value
print(f"6. Calculated Value: {chi2_stat:.4f} vs Critical Value: {critical_value:.4f}")

# 7. State your statistical decision
if chi2_stat > critical_value:
    decision = "Reject the Null Hypothesis (H0)"
else:
    decision = "Fail to Reject the Null Hypothesis (H0)"

print(f"7. Statistical Decision: {decision}")
```
